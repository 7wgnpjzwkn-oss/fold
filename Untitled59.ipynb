```
import os
import http.server
import socketserver
import threading
import json
import numpy as np
import bpy
from playwright.async_api import async_playwright

# ------------------------------------------------------------------------------
# 1. HELPER FUNCTIONS & LOCAL SERVER SETUP
# ------------------------------------------------------------------------------
def find_simulator_root(base_dir):
    if os.path.exists(os.path.join(base_dir, "index.html")):
        return base_dir
    for root, dirs, files in os.walk(base_dir):
        if "index.html" in files:
            return root
    return base_dir

def start_local_server(directory, port=8000):
    class Handler(http.server.SimpleHTTPRequestHandler):
        def __init__(self, *args, **kwargs):
            super().__init__(*args, directory=directory, **kwargs)
        def log_message(self, format, *args):
            pass

    socketserver.TCPServer.allow_reuse_address = True
    server = socketserver.TCPServer(("127.0.0.1", port), Handler)
    thread = threading.Thread(target=server.serve_forever, daemon=True)
    thread.start()
    return server

def setup_two_sided_origami_material(obj_name, front_img_path, back_img_path):
    """
    Creates a material in Blender using Backfacing normal detection to show
    front_img_path on the top face and back_img_path on the bottom face.
    """
    obj = bpy.data.objects.get(obj_name)
    if not obj:
        print(f"[Material Error] Object '{obj_name}' not found!")
        return

    # Select object & enter Edit mode safely
    bpy.context.view_layer.objects.active = obj
    bpy.ops.object.mode_set(mode='EDIT')

    # Select all MESH elements (replaces failing uv.select_all)
    bpy.ops.mesh.select_all(action='SELECT')

    # Standard UV Unwrap
    bpy.ops.uv.unwrap(method='ANGLE_BASED', margin=0.001)

    # Return to Object mode
    bpy.ops.object.mode_set(mode='OBJECT')

    # Create Material
    mat_name = "Origami_TwoSided_Mat"
    mat = bpy.data.materials.get(mat_name)
    if not mat:
        mat = bpy.data.materials.new(name=mat_name)

    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    # 1. Material Output
    output_node = nodes.new(type='ShaderNodeOutputMaterial')
    output_node.location = (800, 0)

    # 2. Mix Shader
    mix_node = nodes.new(type='ShaderNodeMixShader')
    mix_node.location = (550, 0)

    # 3. Geometry Node (Backfacing factor)
    geom_node = nodes.new(type='ShaderNodeNewGeometry')
    geom_node.location = (300, 300)

    # 4. Front BSDF & Image
    bsdf_front = nodes.new(type='ShaderNodeBsdfPrincipled')
    bsdf_front.location = (200, 100)

    img_front = nodes.new(type='ShaderNodeTexImage')
    img_front.location = (-150, 100)
    if front_img_path and os.path.exists(front_img_path):
        img_front.image = bpy.data.images.load(front_img_path)
        print(f"[Material] Loaded Front Image: {front_img_path}")
    else:
        print(f"[Material Warning] Front image path not found: {front_img_path}")

    # 5. Back BSDF, Image & Mapping (Horizontally inverted to prevent backface mirroring)
    bsdf_back = nodes.new(type='ShaderNodeBsdfPrincipled')
    bsdf_back.location = (200, -250)

    img_back = nodes.new(type='ShaderNodeTexImage')
    img_back.location = (-150, -250)
    if back_img_path and os.path.exists(back_img_path):
        img_back.image = bpy.data.images.load(back_img_path)
        print(f"[Material] Loaded Back Image: {back_img_path}")
    else:
        print(f"[Material Warning] Back image path not found: {back_img_path}")

    uv_coord = nodes.new(type='ShaderNodeTexCoord')
    uv_coord.location = (-600, -250)

    mapping = nodes.new(type='ShaderNodeMapping')
    mapping.location = (-380, -250)
    mapping.inputs['Scale'].default_value[0] = -1.0  # Flip X to correct backface view

    # Connect Backface Mapping
    links.new(uv_coord.outputs['UV'], mapping.inputs['Vector'])
    links.new(mapping.outputs['Vector'], img_back.inputs['Vector'])

    # Connect Image Textures to BSDF Base Color
    links.new(img_front.outputs['Color'], bsdf_front.inputs['Base Color'])
    links.new(img_back.outputs['Color'], bsdf_back.inputs['Base Color'])

    # Connect BSDFs to Mix Shader
    links.new(bsdf_front.outputs['BSDF'], mix_node.inputs[1])
    links.new(bsdf_back.outputs['BSDF'], mix_node.inputs[2])

    # Connect Geometry Backfacing to Mix Factor
    links.new(geom_node.outputs['Backfacing'], mix_node.inputs['Factor'])

    # Connect Mix Shader to Material Output
    links.new(mix_node.outputs['Shader'], output_node.inputs['Surface'])

    # Assign material to object
    if obj.data.materials:
        obj.data.materials[0] = mat
    else:
        obj.data.materials.append(mat)

    print(f"[Material] Applied two-sided shader to '{obj_name}'!")

# ------------------------------------------------------------------------------
# 2. MAIN SIMULATION & ANIMATION PIPELINE
# ------------------------------------------------------------------------------
async def run_working_origami_pipeline(fold_file_path, base_simulator_dir, front_jpg_path, back_jpg_path, num_steps=20, port=8000):
    root_dir = find_simulator_root(base_simulator_dir)
    server = start_local_server(root_dir, port)

    captured_frames = []
    faces_data = []

    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch(
                headless=True,
                args=["--disable-web-security", "--allow-file-access-from-files", "--no-sandbox"]
            )
            page = await browser.new_page()

            print(f"[Simulator] Loading page from http://127.0.0.1:{port}/index.html ...")
            await page.goto(f"http://127.0.0.1:{port}/index.html")
            await page.wait_for_load_state("networkidle")

            print(f"[Simulator] Injecting FOLD file: {os.path.basename(fold_file_path)}")
            file_input = page.locator("#fileSelector")
            await file_input.set_input_files(fold_file_path)

            for _ in range(50):
                await page.wait_for_timeout(100)
                ready = await page.evaluate("() => typeof globals !== 'undefined' && globals.model && typeof globals.model.getNodes === 'function' && globals.model.getNodes().length > 0")
                if ready:
                    break

            print("[Simulator] Model successfully loaded into engine!")

            get_positions_js = """
            () => {
                const nodes = globals.model.getNodes();
                return nodes.map(n => {
                    const p = typeof n.getPosition === 'function' ? n.getPosition() : n.position;
                    return [p.x, p.y, p.z];
                });
            }
            """

            faces_data = await page.evaluate("""
            () => {
                const m = globals.model;
                const faces = typeof m.getFaces === 'function' ? m.getFaces() : m.faces;
                if (!faces) return [];
                return faces.map(f => {
                    const nodes = f.nodes || f;
                    return nodes.map(n => typeof n === 'number' ? n : (typeof n.getIndex === 'function' ? n.getIndex() : n.index));
                });
            }
            """)

            # PHASE 1: RESET TO 0% FLAT
            print("\n[Simulator] PHASE 1: Resetting fold to 0% flat...")
            await page.evaluate("""
            () => {
                globals.creasePercent = 0;
                globals.shouldChangeCreasePercent = true;

                if (typeof globals.setCreasePercent === 'function') {
                    try { globals.setCreasePercent(0); } catch(e) {}
                }

                const m = globals.model;
                if (m) {
                    if (typeof m.sync === 'function') m.sync();
                    if (typeof m.syncSolver === 'function') m.syncSolver();
                    if (typeof m.step === 'function') {
                        for (let i = 0; i < 300; i++) m.step();
                    }
                }
            }
            """)

            frame0 = await page.evaluate(get_positions_js)
            p0 = np.array(frame0)
            y_span0 = p0[:, 1].max() - p0[:, 1].min()

            if y_span0 > 0.05:
                print(f"[Simulator] Solver Y-span is {y_span0:.4f}. Using 2D pattern coords for Frame 0 baseline...")
                frame0 = await page.evaluate("""
                () => {
                    const nodes = globals.model.getNodes();
                    return nodes.map(n => {
                        const orig = n._originalPosition || n.originalPosition;
                        return [orig.x, orig.y, orig.z || 0.0];
                    });
                }
                """)
                p0 = np.array(frame0)
                y_span0 = p0[:, 1].max() - p0[:, 1].min()

            captured_frames.append(frame0)
            print(f" -> Frame 0 captured! Base Y-span: {y_span0:.6f}")

            # PHASE 2: STEPPING FOLD PERCENTAGE WITH SOLVER SYNC
            print("\n[Simulator] PHASE 2: Stepping fold percentage with solver sync...")
            TARGET_MAX_FOLD = 0.85
            SOLVER_STEPS_PER_FRAME = 250

            for s in range(1, num_steps + 1):
                target_pct = (s / num_steps) * TARGET_MAX_FOLD

                step_diag = await page.evaluate("""
                (args) => {
                    const pct = args.pct;
                    const steps = args.steps;
                    const m = globals.model;
                    const creases = m.getCreases ? m.getCreases() : (m.creases || []);

                    globals.creasePercent = pct;
                    globals.shouldChangeCreasePercent = true;

                    if (typeof globals.setCreasePercent === 'function') {
                        try { globals.setCreasePercent(pct); } catch(e) {}
                    }

                    if (m) {
                        if (typeof m.sync === 'function') m.sync();
                        if (typeof m.syncSolver === 'function') m.syncSolver();
                    }

                    const c0_theta_after = creases.length > 0 ? creases[0].targetTheta : null;

                    if (m && typeof m.step === 'function') {
                        for (let i = 0; i < steps; i++) {
                            m.step();
                        }
                    }

                    if (typeof globals.render === 'function') globals.render();

                    return {
                        crease0_targetTheta: c0_theta_after
                    };
                }
                """, {"pct": target_pct, "steps": SOLVER_STEPS_PER_FRAME})

                frame = await page.evaluate(get_positions_js)
                captured_frames.append(frame)

                p_curr = np.array(frame)
                disp = np.linalg.norm(p_curr - p0)

                c0_theta_str = f"{step_diag['crease0_targetTheta']:.4f}" if step_diag['crease0_targetTheta'] is not None else "N/A"
                print(f" -> Step {s:02d}/{num_steps} (Fold {target_pct*100:.1f}%): "
                      f"Disp = {disp:.4f} | Crease0 targetTheta = {c0_theta_str}")

            await browser.close()

    finally:
        server.shutdown()
        server.server_close()

    # 3. BLENDER MESH, KEYFRAMES & TWO-SIDED MATERIAL SETUP
    print("\n[Blender] Constructing animated mesh in scene...")
    mesh_name = "Origami_FlappingBird"
    obj_name = "FlappingBird"

    if obj_name in bpy.data.objects:
        bpy.data.objects.remove(bpy.data.objects[obj_name], do_unlink=True)

    mesh = bpy.data.meshes.new(mesh_name)
    obj = bpy.data.objects.new(obj_name, mesh)
    bpy.context.collection.objects.link(obj)

    verts_f0 = captured_frames[0]
    mesh.from_pydata(verts_f0, [], faces_data)
    mesh.update()

    # Keyframe vertex animations across steps
    for step_idx, frame_verts in enumerate(captured_frames):
        frame_number = step_idx * 5 + 1
        bpy.context.scene.frame_set(frame_number)

        for i, v in enumerate(mesh.vertices):
            v.co = frame_verts[i]
            v.keyframe_insert(data_path="co", index=-1)

    print(f"[Blender] Successfully keyframed {len(captured_frames)} animation steps into '{obj_name}'!")

    # Apply Two-Sided Image Material
    print("\n[Blender] Building two-sided material and UV mapping...")
    setup_two_sided_origami_material(obj_name, front_jpg_path, back_jpg_path)

# ------------------------------------------------------------------------------
# EXECUTION
# ------------------------------------------------------------------------------
FOLD_FILE   = "/Users/dobbinp1/Downloads/flappingBird.fold"
ORIGAMI_DIR = "/Users/dobbinp1/Downloads/OrigamiSimulator-main"
FRONT_JPG   = "/Users/dobbinp1/Downloads/IMGP7022.jpg"
BACK_JPG    = "/Users/dobbinp1/Downloads/IMGP7022.jpg"

await run_working_origami_pipeline(
    fold_file_path=FOLD_FILE,
    base_simulator_dir=ORIGAMI_DIR,
    front_jpg_path=FRONT_JPG,
    back_jpg_path=BACK_JPG,
    num_steps=20,
    port=8000
)
```

```
import os
import http.server
import socketserver
import threading
import json
import numpy as np
import bpy
from playwright.async_api import async_playwright

# ------------------------------------------------------------------------------
# 1. HELPER FUNCTIONS & LOCAL SERVER SETUP
# ------------------------------------------------------------------------------
def find_simulator_root(base_dir):
    if os.path.exists(os.path.join(base_dir, "index.html")):
        return base_dir
    for root, dirs, files in os.walk(base_dir):
        if "index.html" in files:
            return root
    return base_dir

def start_local_server(directory, port=8000):
    class Handler(http.server.SimpleHTTPRequestHandler):
        def __init__(self, *args, **kwargs):
            super().__init__(*args, directory=directory, **kwargs)
        def log_message(self, format, *args):
            pass

    socketserver.TCPServer.allow_reuse_address = True
    server = socketserver.TCPServer(("127.0.0.1", port), Handler)
    thread = threading.Thread(target=server.serve_forever, daemon=True)
    thread.start()
    return server

# ------------------------------------------------------------------------------
# 2. MAIN SIMULATION & ANIMATION PIPELINE
# ------------------------------------------------------------------------------
async def run_working_origami_pipeline(fold_file_path, base_simulator_dir, num_steps=20, port=8000):
    root_dir = find_simulator_root(base_simulator_dir)
    server = start_local_server(root_dir, port)

    captured_frames = []
    faces_data = []

    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch(
                headless=True,
                args=["--disable-web-security", "--allow-file-access-from-files", "--no-sandbox"]
            )
            page = await browser.new_page()

            print(f"[Simulator] Loading page from http://127.0.0.1:{port}/index.html ...")
            await page.goto(f"http://127.0.0.1:{port}/index.html")
            await page.wait_for_load_state("networkidle")

            print(f"[Simulator] Injecting FOLD file: {os.path.basename(fold_file_path)}")
            file_input = page.locator("#fileSelector")
            await file_input.set_input_files(fold_file_path)

            # Wait for model load
            for _ in range(50):
                await page.wait_for_timeout(100)
                ready = await page.evaluate("() => typeof globals !== 'undefined' && globals.model && typeof globals.model.getNodes === 'function' && globals.model.getNodes().length > 0")
                if ready:
                    break

            print("[Simulator] Model successfully loaded into engine!")

            # Helper script to pull current 3D vertex positions
            get_positions_js = """
            () => {
                const nodes = globals.model.getNodes();
                return nodes.map(n => {
                    const p = typeof n.getPosition === 'function' ? n.getPosition() : n.position;
                    return [p.x, p.y, p.z];
                });
            }
            """

            # Extract mesh faces
            faces_data = await page.evaluate("""
            () => {
                const m = globals.model;
                const faces = typeof m.getFaces === 'function' ? m.getFaces() : m.faces;
                if (!faces) return [];
                return faces.map(f => {
                    const nodes = f.nodes || f;
                    return nodes.map(n => typeof n === 'number' ? n : (typeof n.getIndex === 'function' ? n.getIndex() : n.index));
                });
            }
            """)

            # ------------------------------------------------------------------
            # PHASE 1: RESET TO 0% FLAT VIA ENGINE FLAGS & SOLVER SYNC
            # ------------------------------------------------------------------
            print("\n[Simulator] PHASE 1: Resetting fold to 0% flat...")
            await page.evaluate("""
            () => {
                globals.creasePercent = 0;
                globals.shouldChangeCreasePercent = true;

                if (typeof globals.setCreasePercent === 'function') {
                    try { globals.setCreasePercent(0); } catch(e) {}
                }

                const m = globals.model;
                if (m) {
                    if (typeof m.sync === 'function') m.sync();
                    if (typeof m.syncSolver === 'function') m.syncSolver();
                    if (typeof m.step === 'function') {
                        for (let i = 0; i < 300; i++) m.step();
                    }
                }
            }
            """)

            frame0 = await page.evaluate(get_positions_js)
            p0 = np.array(frame0)
            y_span0 = p0[:, 1].max() - p0[:, 1].min()

            if y_span0 > 0.05:
                print(f"[Simulator] Solver Y-span is {y_span0:.4f}. Using 2D pattern coords for Frame 0 baseline...")
                frame0 = await page.evaluate("""
                () => {
                    const nodes = globals.model.getNodes();
                    return nodes.map(n => {
                        const orig = n._originalPosition || n.originalPosition;
                        return [orig.x, orig.y, orig.z || 0.0];
                    });
                }
                """)
                p0 = np.array(frame0)
                y_span0 = p0[:, 1].max() - p0[:, 1].min()

            captured_frames.append(frame0)
            print(f" -> Frame 0 captured! Base Y-span: {y_span0:.6f}")

            # ------------------------------------------------------------------
            # PHASE 2: STEPPING FOLD PERCENTAGE WITH EXPLICIT SOLVER SYNC
            # ------------------------------------------------------------------
            print("\n[Simulator] PHASE 2: Stepping fold percentage with solver sync...")
            TARGET_MAX_FOLD = 0.85
            SOLVER_STEPS_PER_FRAME = 250

            for s in range(1, num_steps + 1):
                target_pct = (s / num_steps) * TARGET_MAX_FOLD

                step_diag = await page.evaluate("""
                (args) => {
                    const pct = args.pct;
                    const steps = args.steps;
                    const m = globals.model;
                    const creases = m.getCreases ? m.getCreases() : (m.creases || []);

                    // 1. Update global state variables & flag
                    globals.creasePercent = pct;
                    globals.shouldChangeCreasePercent = true;

                    if (typeof globals.setCreasePercent === 'function') {
                        try { globals.setCreasePercent(pct); } catch(e) {}
                    }

                    // 2. FORCE SIMULATOR & SOLVER SYNC
                    if (m) {
                        if (typeof m.sync === 'function') m.sync();
                        if (typeof m.syncSolver === 'function') m.syncSolver();
                    }

                    const c0_theta_after = creases.length > 0 ? creases[0].targetTheta : null;

                    // 3. Step physics solver
                    if (m && typeof m.step === 'function') {
                        for (let i = 0; i < steps; i++) {
                            m.step();
                        }
                    }

                    if (typeof globals.render === 'function') globals.render();

                    return {
                        crease0_targetTheta: c0_theta_after
                    };
                }
                """, {"pct": target_pct, "steps": SOLVER_STEPS_PER_FRAME})

                frame = await page.evaluate(get_positions_js)
                captured_frames.append(frame)

                p_curr = np.array(frame)
                disp = np.linalg.norm(p_curr - p0)

                c0_theta_str = f"{step_diag['crease0_targetTheta']:.4f}" if step_diag['crease0_targetTheta'] is not None else "N/A"
                print(f" -> Step {s:02d}/{num_steps} (Fold {target_pct*100:.1f}%): "
                      f"Disp = {disp:.4f} | Crease0 targetTheta = {c0_theta_str}")

            await browser.close()

    finally:
        server.shutdown()
        server.server_close()

    # --------------------------------------------------------------------------
    # 3. BLENDER MESH & KEYFRAME GENERATION
    # --------------------------------------------------------------------------
    print("\n[Blender] Constructing animated mesh in scene...")
    mesh_name = "Origami_FlappingBird"
    obj_name = "FlappingBird"

    if obj_name in bpy.data.objects:
        bpy.data.objects.remove(bpy.data.objects[obj_name], do_unlink=True)

    mesh = bpy.data.meshes.new(mesh_name)
    obj = bpy.data.objects.new(obj_name, mesh)
    bpy.context.collection.objects.link(obj)

    verts_f0 = captured_frames[0]
    mesh.from_pydata(verts_f0, [], faces_data)
    mesh.update()

    for step_idx, frame_verts in enumerate(captured_frames):
        frame_number = step_idx * 5 + 1
        bpy.context.scene.frame_set(frame_number)

        for i, v in enumerate(mesh.vertices):
            v.co = frame_verts[i]
            v.keyframe_insert(data_path="co", index=-1)

    print(f"[Blender] Successfully keyframed {len(captured_frames)} animation steps into '{obj_name}'!")

# ------------------------------------------------------------------------------
# EXECUTION
# ------------------------------------------------------------------------------
FOLD_FILE = "/Users/dobbinp1/Downloads/flappingBird.fold"
ORIGAMI_DIR = "/Users/dobbinp1/Downloads/OrigamiSimulator-main"

await run_working_origami_pipeline(FOLD_FILE, ORIGAMI_DIR, num_steps=20, port=8000)
```
